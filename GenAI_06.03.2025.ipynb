```python
# Install the necessary libraries
!pip install transformers sentencepiece

# Import required modules
from transformers import MarianMTModel, MarianTokenizer

# Function to load the Helsinki-NLP model
def get_model(src_lang, tgt_lang):
    model_name = f"Helsinki-NLP/opus-mt-{src_lang}-{tgt_lang}"
    tokenizer = MarianTokenizer.from_pretrained(model_name)
    model = MarianMTModel.from_pretrained(model_name)
    return model, tokenizer

# Function to translate text
def translate_text(text, src_lang="en", tgt_lang="fr"):
    model, tokenizer = get_model(src_lang, tgt_lang)

    # Tokenize the input text
    inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True)

    # Generate translation
    translated_tokens = model.generate(**inputs)

    # Decode the output
    translated_text = tokenizer.batch_decode(translated_tokens, skip_special_tokens=True)

    return translated_text[0]

# Example Usage: Translate English to French
source_text = "Hello, how are you?"
translated_text = translate_text(source_text, src_lang="en", tgt_lang="fr")

print(f"Original Text: {source_text}")
print(f"Translated Text: {translated_text}")
```

```python
# Install gTTS (Google Text-to-Speech)
!pip install gtts

# Import required libraries
from gtts import gTTS
import IPython.display as ipd

# Function to convert text to speech
def text_to_speech(text, lang="en"):
    tts = gTTS(text=text, lang=lang, slow=False)  # Convert text to speech
    tts.save("output.mp3")  # Save the audio file
    return "output.mp3"

# Example Usage
text = "Amritha is the best always !"
audio_file = text_to_speech(text, lang="en")

# Play the audio in Google Colab
ipd.display(ipd.Audio(audio_file))
```
