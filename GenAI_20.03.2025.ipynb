```python
import gradio as gr
from nltk import pos_tag, word_tokenize
import nltk

# Ensure necessary NLTK resources are downloaded
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')

# Define the POS tagging function
def pos_tagging(text):
    try:
        tokens = word_tokenize(text)  # Tokenize the input text
        tagged_tokens = pos_tag(tokens)  # Perform POS tagging
        # Format the output as a string for better readability
        formatted_output = "\n".join([f"{word}: {tag}" for word, tag in tagged_tokens])
        return formatted_output
    except Exception as e:
        return f"Error: {str(e)}"

# Create the Gradio interface
iface = gr.Interface(
    fn=pos_tagging,
    inputs=gr.Textbox(lines=5, label="Input Text"),
    outputs="text",
    title="Parts of Speech Tagging",
    description="Enter a sentence to identify parts of speech for each word."
)

# Launch the Gradio interface
iface.launch()
```

```python
import gradio as gr
from transformers import pipeline

# Load multilingual translation model (supports 50+ languages)
translator = pipeline("translation", model="facebook/mbart-large-50-many-to-many-mmt")

# Language code mapping for common languages
LANGUAGE_MAP = {
    "French": "fr_XX",
    "Spanish": "es_XX",
    "German": "de_DE",
    "Hindi": "hi_IN",
    "Chinese": "zh_CN"
}

# Translation function
def translate_text(text, target_lang):
    try:
        result = translator(text, src_lang="en_XX", tgt_lang=LANGUAGE_MAP[target_lang])
        return result[0]['translation_text']
    except Exception as e:
        return f"Error: {str(e)}"

# Create Gradio interface
iface = gr.Interface(
    fn=translate_text,
    inputs=[
        gr.Textbox(lines=3, placeholder="Enter English text...", label="Input"),
        gr.Dropdown(choices=list(LANGUAGE_MAP.keys()), label="Target Language")
    ],
    outputs=gr.Textbox(label="Translation"),
    title="Multilingual Translator",
    description="Translate English text to multiple languages using Facebook's MBART model",
    examples=[
        ["Hello, how are you?", "French"],
        ["This is a demonstration", "Spanish"]
    ]
)

iface.launch()
```

```python
import gradio as gr
from transformers import pipeline

# Initialize summarization pipeline
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

def summarize_text(text):
    try:
        summary = summarizer(text, max_length=150, min_length=30, do_sample=False)
        return summary[0]['summary_text']
    except Exception as e:
        return f"Error: {str(e)}"

iface = gr.Interface(
    fn=summarize_text,
    inputs=gr.Textbox(lines=10, label="Input Text"),
    outputs=gr.Textbox(label="Summary"),
    title="Text Summarization",
    examples=[["The quick brown fox jumps over the lazy dog. Despite appearances, this simple sentence contains all letters of the English alphabet."]]
)

iface.launch()
```

**Question Answering**

```python
import gradio as gr
from transformers import pipeline

qa_pipeline = pipeline("question-answering", model="deepset/roberta-base-squad2")

def answer_question(context, question):
    try:
        result = qa_pipeline(context=context, question=question)
        return f"Answer: {result['answer']}\nConfidence: {result['score']:.2f}"
    except Exception as e:
        return f"Error: {str(e)}"

iface = gr.Interface(
    fn=answer_question,
    inputs=[
        gr.Textbox(lines=5, label="Context"),
        gr.Textbox(lines=2, label="Question")
    ],
    outputs="text",
    title="Question Answering System",
    examples=[[
        "The Eiffel Tower is located in Paris, France. It was constructed in 1889 as the entrance to the World's Fair.",
        "When was the Eiffel Tower built?"
    ]]
)

iface.launch()
```

```python
from transformers import AutoModelForSequenceClassification, AutoTokenizer, AutoConfig
import numpy as np
from scipy.special import softmax

# Load the model and tokenizer
model_path = "cardiffnlp/twitter-roberta-base-sentiment-latest"
tokenizer = AutoTokenizer.from_pretrained(model_path)
model = AutoModelForSequenceClassification.from_pretrained(model_path)
```

```python
def sentiment_analysis(text):
    encoded_input = tokenizer(text, return_tensors='pt')
    output = model(**encoded_input)
    scores_ = output.logits.detach().numpy()[0]
    scores_ = softmax(scores_)
    labels = ['Negative', 'Neutral', 'Positive']
    scores = {label: float(score) for label, score in zip(labels, scores_)}
    return scores
```

```python
import gradio as gr

# Create the Gradio interface
demo = gr.Interface(
    fn=sentiment_analysis,
    inputs=gr.Textbox(placeholder="Write your text here..."),
    outputs="label",
    examples=[
        ["I love this product!"],
        ["The service was terrible."],
        ["It's an average experience."]
    ],
    title="Sentiment Analysis App",
    description="This app classifies text as Positive, Neutral, or Negative."
)

# Launch the app
demo.launch()
```

```python
from langchain.llms import OpenAI
import os

# Set your OpenAI API key
os.environ["OPENAI_API_KEY"] = "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"

def summarize_text(text):
    # Using 'gpt-3.5-turbo' for summarization with the new chat completion method
    llm = OpenAI(model_name="gpt-3.5-turbo", temperature=0.5)

    # Construct the messages for the chat completion
    messages = [{"role": "user", "content": f"Summarize this text: {text}"}]

    # Get the chat completion response
    response = llm.client.chat.completions.create(
        model=llm.model_name,
        messages=messages,
        temperature=llm.temperature,  # Using llm's temperature setting
    )

    # Extract the summary from the response
    summary = response.choices[0].message.content

    return summary

# Example usage
if __name__ == "__main__":
    input_text = "Pasta is a beloved staple of Italian cuisine, enjoyed worldwide for its versatility and comforting taste. Made from durum wheat semolina and water, it comes in various shapes like spaghetti, penne, and fettuccine, each suited for different sauces. Whether served with a rich tomato-based marinara, creamy Alfredo, or a simple garlic and olive oil dressing, pasta adapts to countless flavors. It can be baked, boiled, or stuffed, making it ideal for diverse culinary traditions. Rich in carbohydrates, it provides energy and pairs well with proteins and vegetables for balanced meals. From classic Italian dishes to modern fusion recipes, pasta remains timeless."
    print("Summary:", summarize_text(input_text))
```
